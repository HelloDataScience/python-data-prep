## 현재 주가 수집

### HTTP 요청 실행

```python
# 필요한 모듈을 임포트합니다.
import requests
```

```python
# 상장회사의 종목코드를 설정합니다.
# [참고] '005930'은 삼성전자 보통주의 종목코드입니다.
ticker = '005930'
```

```python
# 요청 URL을 설정합니다.
url = f'https://finance.naver.com/item/main.naver?code={ticker}'
```

```python
# HTTP 요청을 실행하고 res에 할당합니다.
res = requests.get(url=url)
```

### HTTP 응답 확인

```python
# HTTP 응답 상태 코드를 확인합니다.
res.status_code
```

```python
# HTTP 응답 헤더에서 콘텐트 형태를 확인합니다.
res.headers['content-type']
```

```python
# HTTP 응답 바디 문자열을 출력합니다.
print(res.text)
```

```python
# HTTP 요청 URL을 출력합니다.
print(res.url)
```

### HTML 데이터 처리

```python
# 필요한 모듈을 임포트합니다.
from bs4 import BeautifulSoup as BTS
```

```python
# HTTP 응답 바디 문자열을 bs4.BeautifulSoup으로 변환하고 soup에 할당합니다.
soup = BTS(markup=res.text, features='html.parser')
```

```python
# soup을 확인합니다.
soup
```

```python
# soup을 HTML의 계층 구조(들여쓰기)를 반영한 문자열로 변환하여 출력합니다.
print(soup.prettify())
```

### 현재 주가 수집

```python
# 현재 주가 정보를 포함하는 HTML 요소를 모두 선택하고 items에 할당합니다.
items = soup.select(selector='#rate_info_krx > div > p.no_today > em > span')
```

```python
# items의 원소 개수를 확인합니다.
len(items)
```

```python
# items를 확인합니다.
items
```

```python
# items의 첫 번째 원소에서 현재 주가를 반환합니다.
items[0].text
```

### 현재 주가 수집 함수 생성

```python
# 종목코드를 지정하면 현재 주가를 반환하는 함수를 생성합니다.
def stock_price(ticker):
    url = f'https://finance.naver.com/item/main.naver?code={ticker}'
    res = requests.get(url=url)
    soup = BTS(markup=res.text, features='html.parser')
    items = soup.select(selector='#rate_info_krx > div > p.no_today > em > span')
    return items[0].text
```

### 여러 종목의 현재 주가 수집

```python
# 관심 있는 종목코드를 리스트로 생성합니다.
tickers = ['005930', '000660', '402340', '005935', '009150']
```

```python
# 관심 있는 종목코드의 종목명을 리스트로 생성합니다.
cpnames = ['삼성전자', '하이닉스', 'SK스퀘어', '삼전우선', '삼성전기']
```

```python
# 반복문을 실행합니다.
for ticker, cpname in zip(tickers, cpnames):
    price = stock_price(ticker=ticker)
    print(f'{cpname}의 현재 주가는 {price}원입니다.')
```
