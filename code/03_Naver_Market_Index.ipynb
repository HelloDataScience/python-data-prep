## 주요 시장지표 수집

### HTTP 요청 실행

```python
# 필요한 모듈을 임포트합니다.
import requests
```

```python
# 요청 URL을 설정합니다.
url = 'https://finance.naver.com/marketindex/'
```

```python
# HTTP 요청을 실행하고 res에 할당합니다.
res = requests.get(url=url)
```

### HTTP 응답 확인

```python
# HTTP 응답 상태 코드를 확인합니다.
res.status_code
```

```python
# HTTP 응답 헤더에서 콘텐트 형태를 확인합니다.
res.headers['content-type']
```

```python
# HTTP 응답 바디 문자열을 출력합니다.
print(res.text)
```

```python
# HTTP 요청 URL을 출력합니다.
print(res.url)
```

### HTML 데이터 처리

```python
# 필요한 모듈을 임포트합니다.
from bs4 import BeautifulSoup as BTS
```

```python
# HTTP 응답 바디 문자열을 bs4.BeautifulSoup으로 변환하고 soup에 할당합니다.
soup = BTS(markup=res.text, features='html.parser')
```

```python
# soup을 확인합니다.
soup
```

```python
# soup을 HTML의 계층 구조(들여쓰기)를 반영한 문자열로 변환하여 출력합니다.
print(soup.prettify())
```

### 화폐 단위 수집

```python
# 화폐 단위를 포함하는 HTML 요소를 모두 선택하고 items에 할당합니다.
items = soup.select(selector='#exchangeList > li > a > h3 > span.blind')
```

```python
# items의 원소 개수를 확인합니다.
len(items)
```

```python
# items를 확인합니다.
items
```

```python
# 결과를 저장할 빈 리스트를 생성합니다.
item_texts = []
```

```python
# 반복문으로 items의 각 원소에 있는 텍스트를 item_texts에 추가합니다.
for item in items:
    item_texts.append(item.text)
```

```python
# 위 코드를 리스트 컴프리헨션으로 바꾸면 코드가 간결해집니다.
[item.text for item in items]
```

### 기존 클래스를 상속하고 메서드 추가

```python
# bs4.BeautifulSoup 클래스를 상속하고, 텍스트를 리스트로 반환하는 메서드를 추가합니다.
class MyBTS(BTS):
    def html_text(self, css_selector):
        items = self.select(selector=css_selector)
        return [item.text for item in items]
```

```python
# MyBTS 클래스로 soup2를 생성합니다.
soup2 = MyBTS(markup=res.text, features='html.parser')
```

### 시장지표 수집

```python
# 환전 고시 환율을 리스트로 반환합니다.
soup2.html_text(css_selector='#exchangeList > li > a > div > span.value')
```

```python
# 국제 시장 환율을 리스트로 반환합니다.
soup2.html_text(css_selector='#worldExchangeList > li > a > div > span.value')
```
