### [참고] 시리즈의 문자열 처리: str 접근자

```python
# 정규 표현식 실습용 시리즈를 생성합니다.
nums = pd.Series(data=['M:010-2345-6789', 'T:02-345-9876', '001231-7890123'])
```

```python
# nums의 원소에서 처음 두 글자를 잘라낸 문자열 시리즈를 반환합니다.
# [참고] nums.str.slice(start=0, stop=2)를 실행한 결과와 같습니다.
nums.str[:2]
```

```python
# nums의 원소를 특정 패턴으로 분리하여 리스트를 원소로 갖는 시리즈를 반환합니다.
# [참고] pat 매개변수를 생략하면 모든 공백으로 분리합니다.
nums.str.split(pat=':|-')
```

```python
# nums의 원소를 특정 패턴으로 분리한 리스트의 두 번째 원소를 선택합니다.
nums.str.split(pat=':|-').str[1]
```

```python
# nums의 원소에서 특정 패턴 포함 여부를 원소로 갖는 시리즈를 반환합니다.
nums.str.contains(pat='010')
```

```python
# nums의 원소에서 특정 패턴을 원하는 문자열로 변경한 시리즈를 반환합니다.
nums.str.replace(pat='-', repl='')
```

```python
# nums의 원소에서 시작 위치에 있는 숫자를 추출한 시리즈를 반환합니다.
nums.str.extract(pat='(^[0-9]+)')
```

```python
# nums의 원소에서 마지막 위치에 있는 숫자를 추출한 시리즈를 반환합니다.
nums.str.extract(pat='([0-9]+$)')
```

### 조건에 맞는 행 선택: 연속형 변수

```python
# 거래금액이 200억 이상이면 True, 아니면 False인 cond를 생성합니다.
cond = (apt['거래금액'] >= 2_000_000)
```

```python
# 거래금액이 200억 이상인 행을 선택합니다.
apt.loc[cond, :]
```

```python
# 거래금액이 200억 미만이고 층이 65 이상이면 True, 아니면 False인 cond를 생성합니다.
cond = (apt['거래금액'] < 2_000_000) & (apt['층'] >= 65)
```

```python
# 거래금액이 200억 미만이고 층이 65 이상인 행을 선택합니다.
apt.loc[cond, :]
```

### [참고] 시리즈의 비교 연산 메서드 활용법

```python
# 비교 연산 메서드를 실행한 결과가 시리즈이므로 두 조건을 동시에 만족하는 행을 선택할 때 
# 개별 조건을 소괄호로 감싸지 않아도 됩니다.
apt['거래금액'].lt(other=2_000_000) & apt['층'].ge(other=65)
```

```python
# 층이 10 이상 20 이하면 True, 아니면 False인 불리언 시리즈를 반환합니다.
apt['층'].ge(other=10) & apt['층'].le(other=20)
```

```python
# 위 코드를 between 메서드로 대신할 수 있습니다.
apt['층'].between(left=10, right=20, inclusive='both')
```

### 조건에 맞는 행 선택: 범주형 변수

```python
# 시군구가 '강남구'인 행 개수를 확인합니다.
cond = apt['시군구'].eq(other='강남구')
apt.loc[cond, :].shape
```

```python
# 시군구가 '강남구', '서초구' 또는 '송파구'인 행 개수를 확인합니다.
cond1 = apt['시군구'].eq(other='강남구')
cond2 = apt['시군구'].eq(other='서초구')
cond3 = apt['시군구'].eq(other='송파구')
apt.loc[cond1 | cond2 | cond3, :].shape
```

```python
# isin 메서드는 시리즈 원소가 리스트에 있으면 True, 없으면 False를 반환합니다.
cond = apt['시군구'].isin(values=['강남구', '서초구', '송파구'])
apt.loc[cond, :].shape
```

```python
# str.contains 메서드는 시리즈 원소에 패턴 포함 여부를 True 또는 False로 반환합니다.
cond = apt['시군구'].str.contains(pat='(강남|서초|송파)구')
apt.loc[cond, :].shape
```

### 파생변수 생성

```python
# 거래금액을 전용면적으로 나누고 3.3을 곱한 평당금액을 생성합니다.
apt['평당금액'] = apt['거래금액'] / apt['전용면적'] * 3.3
```

```python
# 거래금액을 10000으로 나누어 거래금액의 단위를 만원에서 억원으로 변경합니다.
apt['거래금액'] = apt['거래금액'] / 10_000
```

```python
# 계약년도에서 입주년도를 뺀 경과년수를 생성합니다.
apt['경과년수'] = apt['계약년도'] - apt['입주년도']
```

```python
# apt의 처음 5행을 확인합니다.
# [참고] 데이터프레임에 없는 열이름으로 시리즈를 할당하면 데이터프레임의 오른쪽에 추가됩니다.
apt.head()
```

### 구간화

```python
# 경과년수가 재건축 연한 기준인 30년 이상이면 True, 아니면 False인 cond를 생성합니다.
cond = apt['경과년수'].ge(other=30)
```

```python
# 경과년수가 30년 이상이면 '충족', 아니면 '부족'인 재건축을 생성합니다.
apt['재건축'] = np.where(cond, '충족', '부족')
```

```python
# apt의 처음 5행을 확인합니다.
apt.head()
```

### 열별 자료형 변환

```python
# apt의 열별 자료형을 확인합니다.
apt.dtypes
```

```python
# 계약년도, 계약월, 계약일을 문자열로 각각 변환합니다.
apt['계약년도'] = apt['계약년도'].astype(dtype=str)
apt['계약월'] = apt['계약월'].astype(dtype=str)
apt['계약일'] = apt['계약일'].astype(dtype=str)
```

### 날짜시간형 변수 생성

```python
# 계약년도, 계약월, 계약일의 원소 사이에 하이픈을 추가한 계약일자를 생성합니다.
apt['계약일자'] = apt['계약년도'] + '-' + apt['계약월'] + '-' + apt['계약일']
```

```python
# apt에서 계약일자를 날짜시간형으로 변환합니다.
apt['계약일자'] = pd.to_datetime(arg=apt['계약일자'], errors='coerce')
```

```python
# apt의 열별 자료형을 확인합니다.
apt.dtypes
```

### [참고] 시리즈의 날짜시간형 처리: dt 접근자

```python
# 계약일자에서 요일을 정수형 시리즈로 반환합니다.
# [참고] Monday는 0, Sunday는 6입니다.
apt['계약일자'].dt.dayofweek
```

```python
# 계약일자에서 영문 요일을 문자열 시리즈로 반환합니다.
# [참고] 요일을 'Monday', ..., 'Sunday'로 반환합니다.
apt['계약일자'].dt.day_name()
```

```python
# 계약일자에서 한글 요일을 문자열 시리즈로 반환합니다.
# [참고] locale 매개변수에 'ko_KR'을 지정하면 요일을 '월요일', ..., '일요일'로 반환합니다.
apt['계약일자'].dt.day_name(locale='ko_KR')
```

```python
# 계약일자에서 지정한 날짜시간 포맷의 문자열을 시리즈로 반환합니다.
# [참고] date_format 매개변수에 날짜시간 포맷을 문자열로 지정합니다.
apt['계약일자'].dt.strftime(date_format='%A')
```

### 결측값 처리: 단순대체

```python
# 등기일자의 결측값을 특정 값으로 대체합니다.
apt['등기일자'].fillna(value='1900-01-01')
```

```python
# 처리일수 평균을 계산하고 days_avg에 할당합니다.
days_avg = apt['처리일수'].dt.days.mean()
```
