## Selenium 활용법

### WebDriver 객체 생성

```python
# 필요한 모듈을 임포트합니다.
from selenium import webdriver
```

```python
# 크롬 브라우저를 제어하기 위한 WebDriver 객체를 생성하고 driver에 할당합니다.
driver = webdriver.Chrome()
```

```python
# HTML 요소를 찾을 때, 최대 3초까지 대기하도록 설정합니다.
driver.implicitly_wait(time_to_wait=3)
```

```python
# 타겟 페이지로 이동합니다.
driver.get(url='https://www.naver.com')
```

### 하이퍼링크 클릭

```python
# 필요한 모듈을 임포트합니다.
from selenium.webdriver.common.by import By
```

```python
# 크롬 브라우저에서 링크 텍스트가 '블로그'와 일치하는 첫 번째 HTML 요소를 클릭합니다.
driver.find_element(by=By.LINK_TEXT, value='블로그').click()
```

### 윈도우 이동

```python
# 현재 열려 있는 크롬 브라우저 윈도우의 핸들 리스트를 확인합니다.
driver.window_handles
```

```python
# 현재 작업 대상을 마지막 윈도우로 전환합니다.
driver.switch_to.window(window_name=driver.window_handles[-1])
```

### 명시적 대기

```python
# 필요한 모듈을 임포트합니다.
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
# 최대 3초까지 특정 조건을 기다리는 WebDriverWait 객체를 생성합니다.
wait = WebDriverWait(driver=driver, timeout=3)
```

```python
# 새 탭이 열려 윈도우 개수가 3이 될 때까지 최대 3초까지 기다립니다.
wait.until(method=EC.number_of_windows_to_be(num_windows=3))
```

### 검색어 입력

```python
# 필요한 모듈을 임포트합니다.
from selenium.webdriver.common.keys import Keys
```

```python
# 크롬 브라우저에서 검색창 관련 첫 번째 HTML 요소를 찾아 search_box에 할당합니다.
search_box = driver.find_element(by=By.NAME, value='sectionBlogQuery')
```

```python
# search_box에 검색어를 입력하고 엔터 키를 누릅니다.
search_box.send_keys('제주여행', Keys.ENTER)
```

### 페이지 이동 & 화면 새로고침

```python
# 이전 페이지로 이동합니다.(뒤로 가기)
driver.back()
```

```python
# 다음 페이지로 이동합니다.(앞으로 가기)
driver.forward()
```

```python
# 현재 페이지를 새로고침합니다.
driver.refresh()
```

```python
# 현재 브라우저 창에 보이는 화면을 png 파일로 저장합니다.
driver.save_screenshot(filename='test.png')
```

### 스크롤 이동: 상대위치

```python
# 현재 스크롤 위치에서 아래로 500 픽셀 이동합니다.
driver.execute_script(script='window.scrollBy(0, 500);')
```

```python
# 현재 스크롤 위치에서 위로 300 픽셀 이동합니다.
driver.execute_script(script='window.scrollBy(0, -300);')
```

### 스크롤 이동: 절대위치

```python
# 현재 웹 페이지의 전체 문서 높이를 반환하는 JavaScript 코드를 작성합니다.
scroll_height_js = 'return document.documentElement.scrollHeight;'
```

```python
# 현재 웹 페이지의 전체 문서 높이를 document_height에 할당합니다.
document_height = driver.execute_script(script=scroll_height_js)
```

```python
# 스크롤을 웹 페이지의 맨 아래로 내립니다.
driver.execute_script(script=f'window.scrollTo(0, {document_height});')
```

```python
# 스크롤을 웹 페이지의 맨 위로 올립니다.
driver.execute_script(script='window.scrollTo(0, 0);')
```

### 첫 번째 블로그로 이동

```python
# 크롬 브라우저에서 첫 번째 블로그 링크를 클릭합니다.
driver.find_element(by=By.PARTIAL_LINK_TEXT, value='제주여행').click()
```

```python
# 현재 작업 대상을 마지막 윈도우로 전환합니다.
driver.switch_to.window(window_name=driver.window_handles[-1])
```

### iframe으로 이동

```python
# HTML 요소에서 iframe 태그를 모두 찾아 items에 할당합니다.
items = driver.find_elements(by=By.TAG_NAME, value='iframe')
```

```python
# items의 원소 개수를 확인합니다.
len(items)
```

```python
# 현재 작업 대상을 지정한 iframe으로 전환합니다.
driver.switch_to.frame(frame_reference='mainFrame')
```

### 텍스트 수집

```python
# 블로그 본문 관련 CSS Selector(선택자)를 css_selector에 할당합니다.
css_selector = 'div.se-module > p > span'
```

```python
# 블로그 본문을 포함하는 모든 HTML 요소를 찾아 items에 할당합니다.
items = driver.find_elements(by=By.CSS_SELECTOR, value=css_selector)
```

```python
# items의 원소 개수를 확인합니다.
len(items)
```

```python
# 리스트 컴프리헨션으로 items의 원소별 텍스트를 리스트로 생성합니다.
[item.text for item in items]
```

### HTML 텍스트 처리

```python
# 필요한 모듈을 임포트합니다.
from bs4 import BeautifulSoup as BTS
```

```python
# 현재 페이지의 HTML 문자열을 bs4.BeautifulSoup으로 변환하고 soup에 할당합니다.
soup = BTS(markup=driver.page_source, features='html.parser')
```

```python
# 블로그 본문을 포함하는 HTML 요소를 모두 선택하고 items에 할당합니다.
items = soup.select(selector=css_selector)
```

```python
# items의 원소별 텍스트를 리스트로 생성하고 하나의 문자열로 결합합니다.
' '.join([item.text for item in items])
```

### 윈도우 닫고 웹 드라이버 종료

```python
# 새 탭을 열고 해당 윈도우로 전환합니다.
driver.switch_to.new_window(type_hint='tab')
```

```python
# 현재 작업 대상인 윈도우를 하나만 닫습니다.
driver.close()
```

```python
# 모든 윈도우를 닫고 WebDriver 세션을 종료합니다.
driver.quit()
```
