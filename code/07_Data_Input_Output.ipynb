### [참고] 특정 문자열을 포함하는 파일 삭제

```python
# 현재 작업 경로에서 'test'를 포함하는 파일명 리스트를 생성합니다.
filenames = [filename for filename in sorted(os.listdir()) if 'test' in filename]
```

```python
# 반복문으로 filenames에 포함된 파일을 모두 삭제합니다.
for filename in filenames:
    os.remove(path=filename)
```

```python
# 현재 작업 경로에 있는 폴더명과 파일명을 확인합니다.
sorted(os.listdir())
```
