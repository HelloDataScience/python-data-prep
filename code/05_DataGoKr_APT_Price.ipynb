### HTTP 요청 실행

```python
# 필요한 모듈을 임포트합니다.
import requests
```

```python
# 요청 URL을 설정합니다.
url = 'https://apis.data.go.kr/1613000/RTMSDataSvcAptTrade/getRTMSDataSvcAptTrade'
```

```python
# 공공데이터포털 오픈 API 인증키를 설정합니다.
api_key = os.getenv(key='DATA_GO_KR_API_KEY')
```

```python
# 요청 변수를 딕셔너리로 생성합니다.
query = dict(LAWD_CD='11110', 
             DEAL_YMD='201512', 
             serviceKey=api_key, 
             pageNo=1, 
             numOfRows=10)
```

```python
# HTTP 요청을 실행하고 res에 할당합니다.
res = requests.get(url=url, params=query)
```

### HTTP 응답 확인

```python
# HTTP 응답 상태 코드를 확인합니다.
res.status_code
```

```python
# HTTP 응답 헤더에서 콘텐트 형태를 확인합니다.
res.headers['content-type']
```

```python
# HTTP 응답 바디 문자열을 출력합니다.
print(res.text)
```

```python
# HTTP 요청 URL을 출력합니다.
print(res.url)
```

### XML 데이터 처리

```python
# 필요한 모듈을 임포트합니다.
import xmltodict
from pprint import pprint
```

```python
# HTTP 응답 바디 문자열을 딕셔너리로 변환하고 data_dict에 할당합니다.
data_dict = xmltodict.parse(xml_input=res.text)
```

```python
# data_dict(중첩된 딕셔너리)의 키와 값을 정렬하여 출력합니다.
pprint(data_dict)
```

### 데이터프레임 생성

```python
# 필요한 모듈을 임포트합니다.
import pandas as pd
```

```python
# data_dict로 데이터프레임을 생성하고 apt_df에 할당합니다.
apt_df = pd.DataFrame(data=data_dict['response']['body']['items']['item'])
```

```python
# apt_df의 처음 5행을 확인합니다.
apt_df.head()
```
