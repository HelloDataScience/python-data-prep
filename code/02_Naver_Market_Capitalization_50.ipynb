## 시가총액 상위 50위 현재가 수집

### HTTP 요청 실행

```python
# 필요한 모듈을 임포트합니다.
import requests
```

```python
# 요청 URL을 설정합니다.
url = 'https://finance.naver.com/sise/sise_market_sum.naver?&page=1'
```

```python
# HTTP 요청을 실행하고 res에 할당합니다.
res = requests.get(url=url)
```

### HTTP 응답 확인

```python
# HTTP 응답 상태 코드를 확인합니다.
res.status_code
```

```python
# HTTP 응답 헤더에서 콘텐트 형태를 확인합니다.
res.headers['content-type']
```

```python
# HTTP 응답 바디 문자열을 출력합니다.
print(res.text)
```

```python
# HTTP 요청 URL을 출력합니다.
print(res.url)
```

### HTML 데이터 처리

```python
# 필요한 모듈을 임포트합니다.
from bs4 import BeautifulSoup as BTS
```

```python
# HTTP 응답 바디 문자열을 bs4.BeautifulSoup으로 변환하고 soup에 할당합니다.
soup = BTS(markup=res.text, features='html.parser')
```

```python
# soup을 확인합니다.
soup
```

```python
# soup을 HTML의 계층 구조(들여쓰기)를 반영한 문자열로 변환하여 출력합니다.
print(soup.prettify())
```

### 시가총액 상위 50위 현재가 수집

```python
# 현재가를 포함하는 HTML 요소를 모두 선택하고 items에 할당합니다.
items = soup.select(selector='td.number:nth-child(3)')
```

```python
# items의 원소 개수를 확인합니다.
len(items)
```

```python
# items를 확인합니다.
items
```

```python
# items의 첫 번째 원소에서 현재 주가를 반환합니다.
items[0].text
```

```python
# 결과를 저장할 빈 리스트를 생성합니다.
item_texts = []
```

```python
# 반복문으로 items의 각 원소에 있는 텍스트를 item_texts에 추가합니다.
for item in items:
    item_texts.append(item.text)
```

```python
# 위 코드를 리스트 컴프리헨션으로 바꾸면 코드가 간결해집니다.
[item.text for item in items]
```

### 시가총액 상위 50위 종목명 수집

```python
# 종목명을 포함하는 HTML 요소를 모두 선택하고 items에 할당합니다.
items = soup.select(selector='td:nth-child(2) > a.tltle')
```

```python
# items의 원소 개수를 확인합니다.
len(items)
```

```python
# items를 확인합니다.
items
```

```python
# items의 각 원소에 있는 텍스트를 원소로 갖는 리스트를 생성합니다.
[item.text for item in items]
```
